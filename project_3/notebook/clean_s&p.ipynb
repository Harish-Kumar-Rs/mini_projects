```python
financial_cols = [
    'price', 'pe_ratio', 'dividend_yield', 'eps',
    'week_low_52', 'week_high_52',
    'market_cap', 'ebitda',
    'price_to_sales', 'price_to_book'
]

df[financial_cols].isnull().all(axis=1).sum()
```

Initial inspection identified 39 companies with all financial variables missing. These records were removed as they contained no usable financial information. The remaining missing values were limited to P/E ratio (24), dividend yield (78), and EBITDA (26), which likely reflect genuine financial characteristics rather than data quality issues.

```python
df = df.dropna(subset=['price'])
```
